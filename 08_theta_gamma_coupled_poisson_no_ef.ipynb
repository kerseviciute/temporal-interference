# LTP → Unsynchronized Theta-Gamma Coupled Poisson inputs (in EF)

```python
synapse_info = pd.read_csv("final_no_ef.csv", index_col = 0)
synapse_info.Delay = 0
synapse_info
```

```python
def generate_poisson_theta_gamma_coupled(
        theta_frequency = 6,
        duration = 1000,
        gamma_rate_base = 100, # base gamma firing rate during theta peak
        fs = 1000 # sampling rate
):
    t = np.arange(0, duration)
    
    # Create theta envelope (modulates gamma)
    theta_phase = 2 * np.pi * theta_frequency * t / 1000
    theta_envelope = (np.sin(theta_phase) + 1) / 2

    spikes = []
    for ti in t:
        gamma_prob = gamma_rate_base / fs * theta_envelope[ti]
        if np.random.rand() < gamma_prob:
            spikes.append(ti)

    spike_times = np.sort(np.array(spikes))
    return spike_times
```

```python
# Parameters
duration = 1 * 1000  # ms # TODO: change to a minute
```

```python
# Generate a list of random seeds to use when generating stimuli
np.random.seed(132)
seeds = np.random.choice(range(10000), 10, replace=False)

seed_counter = -1

# Define the stimulus
def poisson_theta_gamma_coupled():
    global seed_counter
    seed_counter += 1
    
    np.random.seed(seeds[seed_counter])
    
    # Generate spike times
    spike_times = generate_poisson_theta_gamma_coupled(duration = duration)
    spike_vec = h.Vector(spike_times)
    
    stimulus = h.VecStim()
    stimulus.play(spike_vec)
    
    return stimulus
```

```python
neuron = PlasticNeuron(
    synapse_info = synapse_info,
    generate_stimulus = poisson_theta_gamma_coupled
)
```

```python
NeuronUtils.set_stimulus(
    duration = 0
)
```

```python
neuron.run(duration)
```

```python
fig, axs = plt.subplots(nrows = 1, ncols = 2, figsize = (12, 4))

for i, synapse in enumerate(neuron.synapses):
    time = synapse.stimuli
    axs[0].scatter(time, np.repeat(i + 1, repeats = len(time)), color = "black", s = 1)

axs[0].set_ylim(0.5, 10.5)
axs[0].set_ylabel("Neuron")
axs[0].set_yticks(np.arange(1, 11, 1))
axs[0].set_xlabel("Time, ms")
axs[0].set_title("Poisson input firing times")

freqs = []
for synapse in neuron.synapses:
    freqs.append(np.diff(synapse.stimuli)[1:])
freqs = np.concatenate(freqs)

axs[1].hist(freqs, bins = 50)
axs[1].set_xlabel("Time, ms")
axs[1].set_ylabel("Count")
axs[1].set_title(f"Inter-spike interval distribution (mean {round(np.mean(freqs), 2)})")

plt.show()
```
