```python
from plastic_neuron import PlasticNeuron
from abstract_neuron import AbstractNeuron
import matplotlib.pyplot as plt
from neuron_plots import NeuronPlots
from neuron_utils import NeuronUtils
```

# Running the model with a single synapse

```python
neuron = PlasticNeuron(
    n_synapses = 1,
    # Initial conductance of 0.00005 gives ~3 mV of local EPSP at distance 0.5 along the dendrite
    initial_conductance = 0.00005,
    initial_weight = 0,
    generate_delay = lambda: 5,
    generate_stimulus = NeuronUtils.create_constant_freq_stimulus
)
```

```python
neuron.synapse_info
```

```python
NeuronUtils.set_stimulus(duration = 0)
neuron.run(duration = 200)
```

```python
NeuronPlots.plot_all_synapses(neuron)
```

```python
plt.plot(
    neuron.time,
    neuron.voltage,
    linewidth = 0.5,
    color = "black"
)

plt.show()
```
