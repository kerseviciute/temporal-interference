# Running the model with a single synapse

```python
neuron = PlasticNeuron(
    n_synapses = 1,
    # Initial conductance of 0.00005 gives ~3 mV of local EPSP at distance 0.5 along the dendrite
    # Initial conductance of 0.0008 results in combined conductance (through NMDA and AMPA) of 1 nS
    # Initial conductance of 0.0001 results in combined conductance of 0.1 nS
    initial_conductance = 0.0001,
    initial_weight = 0,
    generate_delay = lambda: 5,
    generate_stimulus = NeuronUtils.create_single_stimulus
)
```

```python
neuron.synapse_info
```

```python
NeuronUtils.set_stimulus(duration = 0)
neuron.run(duration = 200)
```
