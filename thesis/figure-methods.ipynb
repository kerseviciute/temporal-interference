```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl
mpl.rcParams["font.family"] = "Avenir"

import pandas as pd
import glob
import re
```

## Subthreshold electric field amplitude

```python
t_start, t_stop, delta_t = 0.1, 1.4, 0.001 # s
time = np.arange(t_start, t_stop, delta_t)
f1 = 10
f2 = 15
delta_f = np.abs(f1 - f2)
phase = np.radians(90)

e1 = - np.sin(2 * np.pi * f1 * time)
e2 = - np.sin(2 * np.pi * f2 * time + phase)

fig, axs = plt.subplots(nrows = 2, ncols = 1, figsize = (3, 3))
fig.subplots_adjust(hspace = -0.25)

axs[0].plot(time, e1, linewidth = 0.75, color = "black")
axs[0].set_ylim(-2.2, 2.2)
axs[0].set_xlim(0.1, 1.4)

for spine_name, spine in axs[0].spines.items():
    spine.set_visible(spine_name == "left")

axs[0].get_xaxis().set_visible(False)
axs[0].set_xticks([])
axs[0].set_yticks([])

axs[1].plot(time, e2, linewidth = 0.75, color = "black")
axs[1].set_ylim(-2.2, 2.2)
axs[1].set_xlim(0.1, 1.4)

for spine_name, spine in axs[1].spines.items():
    spine.set_visible(spine_name == "left")

axs[1].get_xaxis().set_visible(False)
axs[1].set_xticks([])
axs[1].set_yticks([])

# axs[0].set_ylabel(r"$f$", rotation = 0, labelpad = 10, fontsize = 10)
# axs[1].set_ylabel(r"$f + \Delta f$", rotation = 0, labelpad = 21, fontsize = 10)

for ax in axs:
    ax.yaxis.set_label_position("right")
    ax.set_facecolor("none")
    ax.spines["left"].set_color("none")

    ax.set_ylim(-2, 2)

plt.savefig("figure-methods_subthreshold-ef-suprathreshold.png", dpi = 200, bbox_inches = "tight")

for ax in axs:
    ax.set_ylim(-4, 4)

plt.savefig("figure-methods_subthreshold-ef-subthreshold.png", dpi = 200, bbox_inches = "tight")
```

## Traces for subthreshold electric field amplitude

```python
voltage1 = pd.read_csv("../output/temporal-interference-1/figures/figure1/voltage_trace_1000_4Hz_1.csv")
voltage1_1 = pd.read_csv("../output/temporal-interference-1/figures/figure1/voltage_trace_1000_4Hz_1.1.csv")

def plot_voltage_trace(voltage, filename):
	plt.figure(figsize = (4, 2))

	plt.plot(
		voltage.Time / 1000,
		voltage.Voltage,
		color = "black",
		linewidth = 0.5
	)

	plt.ylim(-110, 75)
	plt.ylabel("Voltage, mV")

	plt.xticks([0, 0.5, 1])
	plt.xlabel("Time, s")

	plt.grid(False)

	ax = plt.gca()
	ax.spines["top"].set_visible(False)
	ax.spines["right"].set_visible(False)

	plt.savefig(filename, dpi = 100, bbox_inches = "tight")

plot_voltage_trace(voltage1, filename = "figure-methods_voltage-subthreshold.png")
plot_voltage_trace(voltage1_1, filename = "figure-methods_voltage-suprathreshold.png")
```
