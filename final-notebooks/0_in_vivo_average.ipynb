```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
mean_weight = pd.read_csv("../output/temporal-interference-1/in-vivo/noTI_0.1/mean_synapse_weights.csv")
mean_weight = mean_weight.drop_duplicates(
    subset = "Mean",
    keep = "first",
    ignore_index = True
)
```

```python
mean_weight_0 = pd.read_csv("../output/temporal-interference-1/in-vivo/1000_0_0.32_0.1/mean_synapse_weights.csv")
mean_weight_0 = mean_weight_0.drop_duplicates(
    subset = "Mean",
    keep = "first",
    ignore_index = True
)
```

```python
mean_weight_5 = pd.read_csv("../output/temporal-interference-1/in-vivo/1000_5_0.9_0.1/mean_synapse_weights.csv")
mean_weight_5 = mean_weight_5.drop_duplicates(
    subset = "Mean",
    keep = "first",
    ignore_index = True
)
```

```python
mean_weight_130 = pd.read_csv("../output/temporal-interference-1/in-vivo/1000_130_0.9_0.1/mean_synapse_weights.csv")
mean_weight_130 = mean_weight_130.drop_duplicates(
    subset = "Mean",
    keep = "first",
    ignore_index = True
)
```

```python
weights = {
    "No TI": mean_weight,
    "0 Hz TI": mean_weight_0,
    "5 Hz TI": mean_weight_5,
    "130 Hz TI": mean_weight_130
}
```

```python
colors = {
    "No TI": "tab:blue",
    "0 Hz TI": "tab:red",
    "5 Hz TI": "tab:orange",
    "130 Hz TI": "tab:green"
}
```

```python
fig, ax = plt.subplots(nrows = 1, ncols = 1, figsize = (8, 5))

for condition in weights.keys():
    ax.plot(
        weights[condition].Time / 1000,
        weights[condition].Mean,
        linewidth = 1,
        color = colors[condition],
        label = condition
    )

ax.set_xlabel("Time, s")
ax.set_ylabel("Synaptic weight")

ax.set_ylim(0, 0.4)

ax.grid(alpha = 0.25)

plt.title("Average synaptic weight during naturally occurring inputs", y = 1.1)
plt.legend(loc = "upper center", ncols = 4, frameon = False, bbox_to_anchor = (0.5, 1.1))
plt.show()
```
