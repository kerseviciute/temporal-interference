```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
t_start, t_stop, delta_t = 0.1, 1.4, 0.001 # s
time = np.arange(t_start, t_stop, delta_t)
f1 = 5
f2 = 5
delta_f = np.abs(f1 - f2)
phase = np.radians(90)

e1 = - np.sin(2 * np.pi * f1 * time)
e2 = - np.sin(2 * np.pi * f2 * time + phase)
sum_e = np.sin(2 * np.pi * delta_f / 2 * time - phase / 2) * 2
```

```python
fig, axs = plt.subplots(nrows = 3, ncols = 1, figsize = (4, 6))

axs[0].plot(time, e1, linewidth = 1, color = "black")
axs[0].set_ylim(-2.2, 2.2)
# axs[0].axhline(0, linewidth = 0.5, color = "black", alpha = 0.5)
axs[0].set_xlim(0.1, 1.4)

for spine_name, spine in axs[0].spines.items():
    spine.set_visible(spine_name == "left")

axs[0].get_xaxis().set_visible(False)
axs[0].set_yticks([-1, 0, 1])
axs[0].set_yticklabels([r"$-$I$_1$", "0", r"I$_1$"])

axs[1].plot(time, e2, linewidth = 1, color = "black")
axs[1].set_ylim(-2.2, 2.2)
# axs[1].axhline(0, linewidth = 0.5, color = "black", alpha = 0.5)
axs[1].set_xlim(0.1, 1.4)

for spine_name, spine in axs[1].spines.items():
    spine.set_visible(spine_name == "left")

axs[1].get_xaxis().set_visible(False)
axs[1].set_yticks([-1, 0, 1])
axs[1].set_yticklabels([r"$-$I$_2$", "0", r"I$_2$"])

axs[2].plot(time, e1 + e2, linewidth = 1, color = "black")
axs[2].set_ylim(-2.2, 2.2)

for spine_name, spine in axs[2].spines.items():
    spine.set_visible(spine_name == "left")

axs[2].get_xaxis().set_visible(False)
# axs[2].axhline(0, linewidth = 0.5, color = "black", alpha = 0.5)
axs[2].set_xlim(0.1, 1.4)
axs[2].set_yticks([-2, 0, 2])
axs[2].set_yticklabels([r"$-$(I$_1+$I$_2$)", "0", r"(I$_1+$I$_2$)"])

fig.lines.append(
    plt.Line2D(
        [0.511, 0.511], [0.05, 0.92],
        transform = fig.transFigure,
        color = "black",
        linestyle = (0, (5, 5)),
        linewidth = 0.5
    )
)

fig.lines.append(
    plt.Line2D(
        [0.212, 0.212], [0.05, 0.92],
        transform = fig.transFigure,
        color = "black",
        linestyle = (0, (5, 5)),
        linewidth = 0.5
    )
)

fig.text(
    0.53, 0.88,
    "In phase",
    rotation = 0,
    va = "center",
    ha = "left"
)

fig.text(
    0.23, 0.88,
    "Out of phase",
    rotation = 0,
    va = "center",
    ha = "left"
)

plt.show()
```
