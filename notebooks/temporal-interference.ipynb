```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
t_start, t_stop, delta_t = 0.0, 0.5, 0.00001 # s
time = np.arange(t_start, t_stop, delta_t)
f1 = 50
f2 = 55
delta_f = np.abs(f1 - f2)
phase = np.radians(90)

e1 = np.sin(2 * np.pi * f1 * time)
e2 = np.sin(2 * np.pi * f2 * time + phase)
sum_e = np.sin(2 * np.pi * delta_f / 2 * time - phase / 2) * 2
```

```python
plt.plot(time, e1, label = rf"$f_1 = {f1}$")
plt.plot(time, e2, label = rf"$f_2 = {f2}$")
plt.plot(time, e1 + e2, label = rf"$f_1 + f_2$")
plt.plot(time, sum_e, label = rf"$\Delta f = {delta_f}$")
plt.legend(loc = "center left", bbox_to_anchor = (1, 0.5))
plt.show()
```
