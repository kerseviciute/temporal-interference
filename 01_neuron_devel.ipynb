```python
import pandas as pd
import numpy as np
from plastic_neuron import PlasticNeuron
from abstract_neuron import AbstractNeuron
import matplotlib.pyplot as plt
```

```python
neuron = PlasticNeuron(
    connection_weight = 0.00009,
    initial_weight = 0.00009 * 0.75
)
```

```python
neuron.run(
    duration = 500
)
```

```python
plt.plot(
    neuron.output.Time,
    neuron.output.Soma,
    linewidth = 0.5,
    color = "black"
)

plt.ylim(-110, 70)

plt.show()
```

```python
time = neuron.output.Time
for synapse in neuron.synapses:
    plt.plot(time, synapse.weights)

plt.ylim(0, 1)
plt.show()
```
